```python
import numpy as np
import matplotlib.pyplot as plt
from scipy.stats import lognorm
from scipy.integrate import quad

mu = 0
sigma = 0.5
p = 0.95

dist = lognorm(s=sigma, scale=np.exp(mu))

VaR = dist.ppf(p)

def integrand(x):
    return x * dist.pdf(x)

ES_integral, _ = quad(integrand, VaR, np.inf)
ES = ES_integral / (1 - p)

x = np.linspace(0, 5, 500)
pdf_values = dist.pdf(x)

plt.figure(figsize=(10, 6))
plt.plot(x, pdf_values, 'b-', lw=2, label='')

plt.fill_between(x[x <= VaR], pdf_values[x <= VaR], color='skyblue', alpha=0.5)

plt.axvline(VaR, color='red', linestyle='--', lw=1.5)
plt.text(VaR, -0.005, f'VaR$_p$(X)',
         horizontalalignment='center', verticalalignment='top', size='x-large')

plt.axvline(ES, color='green', linestyle='--', lw=1.5)
plt.text(ES, -0.005, f'ES$_p$(X)',
         horizontalalignment='center', verticalalignment='top', size='x-large')

plt.xlabel('')
plt.ylabel('Плотность вероятности', size = 'x-large')
plt.ylim(bottom=0)

plt.tick_params(
    axis='both',
    which='both',
    bottom=False,
    left=False,
    labelbottom=False,
    labelleft=False
)

plt.grid(False)
plt.legend()
plt.show()
```
