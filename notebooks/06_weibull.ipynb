```python
import math
import matplotlib.pyplot as plt
import numpy as np
import scipy.stats as stats
from scipy.integrate import quad
from scipy.optimize import brentq

def weibull_isf(k, q):
    return stats.weibull_min.isf(q, k)

def eq_for_c(c, k, eps, var):
    def integrand(t):
        p = c * eps * t
        return weibull_isf(k, p)
    integral, _ = quad(integrand, 0, 1, epsabs=1e-14, epsrel=1e-14, limit=1000)
    return integral - var

def find_c(k, eps, var):
    c_sol = brentq(eq_for_c, 1, 1 / eps, args=(k, eps, var), maxiter=10000)
    return c_sol

k = 0.5
eps = 1e-308

var = stats.weibull_min.isf(eps, k)

c = find_c(k, eps, var)

print(f"c = {c}")
```
