```python
import scipy.stats as stats
from scipy.integrate import quad
from scipy.optimize import brentq
import math

def gamma_ppf(alpha, beta, x_0, q):
    return x_0 + stats.gamma.ppf(q, beta, scale=1/alpha)

def eq_for_c(c, alpha, beta, x_0, eps, var):
    lower = 1 - c * eps
    upper = 1
    integral, _ = quad(lambda q: gamma_ppf(alpha, beta, x_0, q), lower, upper)
    left_side = (1 / (c * eps)) * integral
    return left_side - var

def find_c(alpha, beta, x_0, eps, var):
    c_sol = brentq(eq_for_c, 1, 1 / eps, args=(alpha, beta, x_0, eps, var), maxiter=10000, xtol=1e-15)
    return c_sol

alpha, beta, x_0 = 1, 9, 0
eps = 1e-1
var = gamma_ppf(alpha, beta, x_0, 1 - eps)

for _ in range(0, 15):
    var = gamma_ppf(alpha, beta, x_0, 1 - eps)
    c = find_c(alpha, beta, x_0, eps, var)

    L=math.log(eps)
    print(f"{eps:<10.1e} | {c:<8.5f} | {math.e:<8.5f} | {1/L**2:<8.5f}")
    eps /= 10
```
